## Правила разметки pdf -> md

Строки которые начинаются с:


- Раздел I, Раздел II, Раздел III - H1 (#)  - Римская нумерация
- ГЛАВА 1, ГЛАВА 2, ГЛАВА 3  - H2 (##)      - Арапская нумерация
- Статья 1, Статья 2, Статья 3  - H3 (###)  - Арапская нумерация

```python
import fitz  # PyMuPDF
import re

def pdf_to_markdown(pdf_path, output_path):
    doc = fitz.open(pdf_path)
    markdown_content = []
    
    # Регулярные выражения для определения заголовков
    # Раздел с римскими цифрами (H1)
    section_pattern = re.compile(r'^Раздел\s+([IVX]+)', re.IGNORECASE)
    
    # Глава с арабскими цифрами (H2)
    chapter_pattern = re.compile(r'^ГЛАВА\s+(\d+)', re.IGNORECASE)
    
    # Статья с арабскими цифрами (H3)
    article_pattern = re.compile(r'^Статья\s+(\d+)', re.IGNORECASE)
    
    for page_num in range(doc.page_count):
        page = doc[page_num]
        text = page.get_text()
        
        # Basic formatting cleanup
        lines = text.split('\n')
        for line in lines:
            line = line.strip()
            if line:
                # Проверяем заголовки по приоритету (от более специфичных к менее специфичным)
                
                # Проверка на "Раздел" с римскими цифрами (H1)
                if section_pattern.match(line):
                    markdown_content.append('\n')
                    markdown_content.append(f"# {line}")
                    markdown_content.append('\n')
                # Проверка на "ГЛАВА" с арабскими цифрами (H2)
                elif chapter_pattern.match(line):
                    markdown_content.append('\n')
                    markdown_content.append(f"## {line}")
                    markdown_content.append('\n')
                # Проверка на "Статья" с арабскими цифрами (H3)
                elif article_pattern.match(line):
                    markdown_content.append('\n')
                    markdown_content.append(f"### {line}")
                    markdown_content.append('\n')
                else:
                    markdown_content.append(line)
        
        #markdown_content.append("\n---\n")  # Page separator
    
    # Save to file
    with open(output_path, 'w', encoding='utf-8') as f:
        f.write('\n'.join(markdown_content))
    
    doc.close()
```
